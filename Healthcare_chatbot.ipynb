```python
import json
with open('Medical_intent.json',encoding='utf-8') as file:
    data_json=json.load(file)
```

```python
row=[]
for intent in data_json['intents']:
    tag=intent['tag']
    for pattern in intent['patterns']:
        row.append([pattern,tag])
```

```python
import pandas as pd
df=pd.DataFrame(row,columns=['Pattern','Tag'])
df
```

```python
df.to_csv('Health_chatbot.csv')
```

```python
data=pd.read_csv('Health_chatbot.csv')
data.head()
```

```python
X=data['Pattern'].str.lower()
Y=data['Tag']
```

```python
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
token=Tokenizer()
token.fit_on_texts(X)
X=token.texts_to_sequences(X)
X=pad_sequences(X)
```

```python
#instead using label encoder
#y=list(set(Y))
#Y_out=[]
#for i in Y:
 #   Y_out.append(y.index(i))
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
Y=le.fit_transform(Y)
Y
```

```python
#model Building 
model=tf.keras.Sequential([
    layers.Embedding(input_dim=len(token.word_index)+1,output_dim=128),
    layers.LSTM(64),
    layers.Dense(128,activation='relu'),
    layers.Dense(len(le.classes_),activation='softmax')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history=model.fit(X,Y,epochs=400,validation_split=0.2)
```

```python
max_len=X.shape[1]
max_len
```

```python
import numpy as np
def predict(pred):
    pred=pred.lower()
    pred=token.texts_to_sequences([pred])
    pred=pad_sequences(pred,maxlen=max_len)
    Y_pred=model.predict(pred)
    print(Y_pred)
    index=np.argmax(Y_pred)
    tag=le.inverse_transform([index])
    print(tag,index)
predict('hello')
```

```python
model.save('Health_chatbot.h5')
```
